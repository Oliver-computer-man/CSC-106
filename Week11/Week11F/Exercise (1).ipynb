# Generate Data

```python
import numpy as np
x_values = np.linspace(start = -2, stop = 2, num = 41)
```

```python
print(x_values)
```

```python
def relu(x_values):
    return(np.maximum(0, x_values))
```

```python
relu_values = relu(x_values)
```

```python
print(relu_values)
```

```python
import plotly.express as px
px.line(x = x_values, y = relu_values)
```
